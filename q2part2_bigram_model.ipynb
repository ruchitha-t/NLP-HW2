Bigam Language Model

Build Unigram and Bigram Counts

```python
from collections import Counter
```

```python
def build_counts(corpus):

    unigram_counts = Counter()
    bigram_counts = Counter()
    previous_word_counts = Counter()

    for sentence in corpus:

        words = sentence.split()

        # Count unigrams
        for word in words:
            unigram_counts[word] += 1

        # Count bigrams
        for i in range(len(words) - 1):

            previous_word = words[i]
            next_word = words[i + 1]

            bigram_counts[
                (previous_word, next_word)
            ] += 1

            previous_word_counts[previous_word] += 1

    return (
        unigram_counts,
        bigram_counts,
        previous_word_counts
    )
```

Calculate Bigram Probability using MLE

```python
def bigram_probability(
    previous_word,
    next_word,
    bigram_counts,
    previous_word_counts
):

    numerator = bigram_counts[
        (previous_word, next_word)
    ]

    denominator = previous_word_counts[
        previous_word
    ]

    if denominator == 0:
        return 0.0

    return numerator / denominator
```

Calcullate Sentence Probability

```python


def sentence_probability(
    sentence,
    bigram_counts,
    previous_word_counts
):

    words = sentence.split()

    probability = 1.0

    for i in range(len(words) - 1):

        previous_word = words[i]
        next_word = words[i + 1]

        probability *= bigram_probability(
            previous_word,
            next_word,
            bigram_counts,
            previous_word_counts
        )

    return probability
```

Main

```python
if __name__ == "__main__":

    # --------------------------------------------------------
    # Training Corpus
    # --------------------------------------------------------

    corpus = [
        "<s> I love NLP </s>",
        "<s> I love deep learning </s>",
        "<s> deep learning is fun </s>"
    ]

    # --------------------------------------------------------
    # Compute Counts
    # --------------------------------------------------------

    (
        unigram_counts,
        bigram_counts,
        previous_word_counts
    ) = build_counts(corpus)

    # --------------------------------------------------------
    # Print Unigram Counts
    # --------------------------------------------------------

    print("===== Unigram Counts =====")

    for word, count in unigram_counts.items():

        print(f"{word}: {count}")

    # --------------------------------------------------------
    # Print Bigram Counts
    # --------------------------------------------------------

    print("\n===== Bigram Counts =====")

    for (
        previous_word,
        next_word
    ), count in bigram_counts.items():

        print(
            f"({previous_word}, {next_word}): {count}"
        )

    # --------------------------------------------------------
    # Print Bigram Probabilities
    # --------------------------------------------------------

    print("\n===== Bigram Probabilities =====")

    for (
        previous_word,
        next_word
    ), count in bigram_counts.items():

        probability = bigram_probability(
            previous_word,
            next_word,
            bigram_counts,
            previous_word_counts
        )

        print(
            f"P({next_word} | {previous_word}) "
            f"= {probability:.4f}"
        )

    # --------------------------------------------------------
    # Test Required Sentences
    # --------------------------------------------------------

    sentence1 = "<s> I love NLP </s>"

    sentence2 = "<s> I love deep learning </s>"

    probability1 = sentence_probability(
        sentence1,
        bigram_counts,
        previous_word_counts
    )

    probability2 = sentence_probability(
        sentence2,
        bigram_counts,
        previous_word_counts
    )

    print("\n===== Sentence Probabilities =====")

    print(
        f"{sentence1} = "
        f"{probability1:.4f}"
    )

    print(
        f"{sentence2} = "
        f"{probability2:.4f}"
    )

    # --------------------------------------------------------
    # Compare Sentences
    # --------------------------------------------------------

    if probability1 > probability2:

        print("\nPreferred Sentence: Sentence 1")

        print(
            "Reason: Sentence 1 has a higher "
            "bigram probability."
        )

    elif probability2 > probability1:

        print("\nPreferred Sentence: Sentence 2")

        print(
            "Reason: Sentence 2 has a higher "
            "bigram probability."
        )

    else:

        print("\nBoth sentences have the same probability.")
```
