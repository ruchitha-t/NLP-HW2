```python
def calculate_metrics(confusion_matrix):

    classes = ["Cat", "Dog", "Rabbit"]

    precisions = {}
    recalls = {}

    # Calculate precision and recall for each class
    for i, class_name in enumerate(classes):

        # True Positive
        true_positive = confusion_matrix[i][i]

        # Total predictions made for this class
        predicted_total = sum(confusion_matrix[i])

        # Total actual instances of this class
        actual_total = sum(row[i] for row in confusion_matrix)

        # Precision = TP / (TP + FP)
        precision = true_positive / predicted_total

        # Recall = TP / (TP + FN)
        recall = true_positive / actual_total

        precisions[class_name] = precision
        recalls[class_name] = recall

    # --------------------------------------------------------
    # Macro Averaging
    # --------------------------------------------------------

    macro_precision = sum(precisions.values()) / len(classes)

    macro_recall = sum(recalls.values()) / len(classes)

    # --------------------------------------------------------
    # Micro Averaging
    # --------------------------------------------------------

    # Sum of all True Positives
    total_true_positive = sum(
        confusion_matrix[i][i]
        for i in range(len(classes))
    )

    # Total number of predictions
    total_predictions = sum(
        sum(row) for row in confusion_matrix
    )

    micro_precision = total_true_positive / total_predictions

    micro_recall = total_true_positive / total_predictions

    # --------------------------------------------------------
    # Print Results
    # --------------------------------------------------------

    print("===== Confusion Matrix Evaluation =====")

    print("\nPer-Class Precision and Recall:")

    for class_name in classes:

        print(
            f"{class_name}: "
            f"Precision = {precisions[class_name]:.4f}, "
            f"Recall = {recalls[class_name]:.4f}"
        )

    print("\nMacro Averaging:")
    print(f"Macro Precision = {macro_precision:.4f}")
    print(f"Macro Recall    = {macro_recall:.4f}")

    print("\nMicro Averaging:")
    print(f"Micro Precision = {micro_precision:.4f}")
    print(f"Micro Recall    = {micro_recall:.4f}")



if __name__ == "__main__":

    # Confusion Matrix
    #
    # Rows    = System predictions
    # Columns = Gold labels
    #
    #             Cat   Dog   Rabbit
    # Cat          5    10      5
    # Dog         15    20     10
    # Rabbit       0    15     10

    confusion_matrix = [
        [5, 10, 5],
        [15, 20, 10],
        [0, 15, 10]
    ]

    calculate_metrics(confusion_matrix)
```
